# Affine Cipher Technique

## History
The Affine Cipher is a type of monoalphabetic substitution cipher, where each letter in an alphabet is mapped to its numeric equivalent, encrypted using a simple mathematical function, and converted back to a letter. It is more secure than the Caesar Cipher because it uses a linear function involving two keys ($a$ and $b$) instead of just one shift.

## What is Affine Cipher Technique?
The Affine Cipher is a special case of the more general substitution cipher. To encrypt, the algorithm uses modular arithmetic to transform the integer assigned to each character. It requires two keys:
1.  **Key $a$**: The multiplicative key. It must be between $1$ and $N-1$ and satisfy the condition $\gcd(a, N) = 1$.
2.  **Key $b$**: The additive key (shift). It must be between $0$ and $N-1$.

In this implementation, we use the printable ASCII range from space (` `) to tilde (`~`), which spans from ASCII value 32 to 126 ($N=95$).

## Cryptography Algorithm

### Constants
*   $S = 32$ (Start of printable ASCII range)
*   $E = 126$ (End of printable ASCII range)
*   $N = E - S + 1 = 95$ (Total number of printable characters)
*   $a$ = Multiplicative key ($1 \le a < N$, must be coprime with $N$)
*   $b$ = Additive key ($0 \le b < N$)
*   $x$ = Numeric value of the plaintext character ($0 \le x < N$)
*   $y$ = Numeric value of the ciphertext character ($0 \le y < N$)

### 1. Encryption
For each character in the plaintext, the character index $x$ is transformed into a ciphertext index $y$ using the formula:

$$E(x) = (a \cdot x + b) \pmod N$$

To get the final ASCII value: $C = E(x) + S$

### 2. Decryption
To reverse the process, the decryption function uses the modular multiplicative inverse of $a$ (denoted as $a^{-1}$) such that $(a \cdot a^{-1}) \pmod N = 1$:

$$D(y) = a^{-1} \cdot (y - b) \pmod N$$

To get the final ASCII value: $P = D(y) + S$

### Key Requirements and Ranges
*   **Condition for $a$**: $\gcd(a, N) = 1$. This ensures that every plaintext character maps to a unique ciphertext character, making the function invertible.
*   **Range of $a$**: Typically $1 \le a < N$.
*   **Range of $b$**: $0 \le b < N$.

### 1. Import Dependenciess

```python
import random
import math
```

### 2. Halper Utilities

```python
def mod_inverse(key: int, m: int) -> int:
    # find modular inverse of key under modulo m
    for i in range(1, m):
        if (key * i) % m == 1:
            return i
    return None  # no inverse exists if key and m are not coprime
```

### 3. Generate Random Keyes

```python
def generate_random_key():
    TOTAL_CHAR = 126 - 32 + 1  # 95
    while True:
        key = random.randint(2, 1000)
        if math.gcd(key, TOTAL_CHAR) == 1:
            return key
```

### 4. Encryption

```python
def encrypt(text: str, key: int) -> str:
    START_ASCII = 32
    END_ASCII = 126
    TOTAL_CHAR = END_ASCII - START_ASCII + 1  # 95 characters

    if math.gcd(key, TOTAL_CHAR) != 1:
        raise ValueError(f"'key' must be coprime with {TOTAL_CHAR}. Choose a different key.")

    encrypted_text = ""

    for ch in text:
        code = ord(ch)
        if START_ASCII <= code <= END_ASCII:
            cipher_code = ((code - START_ASCII) * key) % TOTAL_CHAR
            encrypted_text += chr(cipher_code + START_ASCII)
        else:
            encrypted_text += ch

    return encrypted_text
```

### 5. Decryption

```python
def decrypt(text: str, key: int) -> str:
    START_ASCII = 32
    END_ASCII = 126
    TOTAL_CHAR = END_ASCII - START_ASCII + 1  # 95 characters

    key_inv = mod_inverse(key, TOTAL_CHAR)
    if key_inv is None:
        raise ValueError(f"'key' has no modular inverse under mod {TOTAL_CHAR}. Choose a coprime key.")

    decrypted_text = ""

    for ch in text:
        code = ord(ch)
        if START_ASCII <= code <= END_ASCII:
            plain_code = ((code - START_ASCII) * key_inv) % TOTAL_CHAR
            decrypted_text += chr(plain_code + START_ASCII)
        else:
            decrypted_text += ch

    return decrypted_text
```

### 6. Example usage

```python
key = generate_random_key()
print(f"Generated Random Key: {key}")
```

```python
plaintext = """TOP secret Massage! Agent 101, visit Area 51 (37d14'0"N 115d48'30"W)."""
print(f"Original Plain Text: {plaintext}")

cipher_text = encrypt(plaintext, key)
print("Encrypted:", cipher_text)

decrypted_text = decrypt(cipher_text, key)
print("Decrypted:", decrypted_text)

match = plaintext == decrypted_text
print(f"Verification Match:{match}")
```
