# Caesar Cipher Technique

## History
The Caesar cipher is a simple encryption technique that was used by Julius Caesar to send secret messages to his allies. It works by shifting the letters in the plaintext message by a certain number of positions, known as the "shift" or "key". The Caesar Cipher technique is one of the earliest and simplest methods of encryption techniques.

## What is Caesar Cipher Technique?
The Caesar Cipher is one of the simplest and most widely known encryption techniques. It is a type of **substitution cipher** in which each letter or character in the plaintext is replaced by a character some fixed number of positions down the alphabet or character set.

In this specific implementation, we are using the printable ASCII range from space (` `) to tilde (`~`), which spans from ASCII value 32 to 126.

## Cryptography Algorithm

### Constants
*   $S = 32$ (Start of printable ASCII range)
*   $E = 126$ (End of printable ASCII range)
*   $N = E - S + 1 = 95$ (Total number of printable characters)
*   $k$ = Shift key
*   $p$ = Numeric value of the plaintext character
*   $c$ = Numeric value of the ciphertext character

### 1. Encryption
For each character in the plaintext, the encryption process transforms the character value $p$ into a ciphertext value $c$ using the following formula:

$$E_k(p) = (((p - S) + k)\mod N) + S$$

### 2. Decryption
To reverse the process, the decryption function transforms the ciphertext value $c$ back into the original plaintext value $p$ using the key $k$:

$$D_k(c) = (((c - S) - k) \mod N) + S$$

### Key Wrapping
Since the shift $k$ can be any integer, we normalize it to ensure it fits within our range:
$$K_{normalized} = k \mod N$$

### 1. Encryption

```python
def encrypt(text: str, key: int) -> str:
    """
    Encrypt text using a Caesar Cipher shift over printable ASCII 32 (' ') - 126 ('~').

    Parameters
    ----------
    text : str
        Text to encrypt.
    key : int
        Shift value. Any integer works, it wraps automatically.

    Returns
    -------
    str
        Encrypted text.
    """

    START_ASCII = 32
    END_ASCII = 126
    TOTAL_CHAR = END_ASCII - START_ASCII + 1
    KEY = key % TOTAL_CHAR

    encrypted_text = ""

    for char in text:
        code = ord(char)
        # check the char is printable
        if START_ASCII <= code <= END_ASCII:
            new_code = (((code - START_ASCII) + KEY) % TOTAL_CHAR) + START_ASCII
            encrypted_text += chr(new_code)
        else:
            encrypted_text += char
    return encrypted_text
```

### 2. Decryption

```python
def decrypt(text: str, key: int) -> str:
    """
    Decrypt text using a Caesar Cipher shift over printable ASCII 32 (' ') - 126 ('~').

    Parameters
    ----------
    text : str
        Text to decrypt.
    key : int
        Shift value used during encryption. Any integer works, it wraps automatically.

    Returns
    -------
    str
        Decrypted (plain) text.
    """

    START_ASCII = 32
    END_ASCII = 126
    TOTAL_CHAR = END_ASCII - START_ASCII + 1
    KEY = key % TOTAL_CHAR

    decrypted_text = ""

    for char in text:
        code = ord(char)
        # check the char is printable
        if START_ASCII <= code <= END_ASCII:
            new_code = (((code - START_ASCII) - KEY) % TOTAL_CHAR) + START_ASCII
            decrypted_text += chr(new_code)
        else:
            decrypted_text += char
    return decrypted_text
```

### 3. Example of Usage

```python
# Configuration
PLAIN_TEXT = """TOP secret Massage! Agent 101, visit Area 51 (37d14'0"N 115d48'30"W)."""
KEY = 13

# Encryption Process
cipher_text = encrypt(PLAIN_TEXT, KEY)
print(f"Original Plain Text:  {PLAIN_TEXT}")
print(f"Encrypted Cipher:     {cipher_text}")

# Decryption Process
decrypted_text = decrypt(cipher_text, KEY)
print(f"Decrypted Text:       {decrypted_text}")

# Verification
match = PLAIN_TEXT == decrypted_text
print(f"Verification Match:   {match}")
```
